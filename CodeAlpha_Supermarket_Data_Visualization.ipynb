```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("supermarket_sales.csv")
```

```python
category_counts = df["Category Name"].value_counts()

print(category_counts)
```

```python
plt.figure(figsize=(10, 6))

category_counts.plot(kind="bar")

plt.title("Number of Items by Category")
plt.xlabel("Category Name")
plt.ylabel("Number of Items")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(8, 8))

df["Category Name"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Category Distribution")
plt.ylabel("")

plt.show()
```

```python
top_items = df.head(10)

plt.figure(figsize=(12, 6))

plt.bar(top_items["Item Name"], top_items["Item Code"])

plt.title("Top 10 Items by Item Code")
plt.xlabel("Item Name")
plt.ylabel("Item Code")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.show()
```

```python
category_summary = df.groupby("Category Name").size().reset_index(name="Item Count")

print(category_summary)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=category_summary,
    x="Category Name",
    y="Item Count"
)

plt.title("Category-wise Item Count")
plt.xlabel("Category Name")
plt.ylabel("Number of Items")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
category_code_counts = df["Category Code"].value_counts()

plt.figure(figsize=(10, 6))

category_code_counts.plot(kind="bar")

plt.title("Number of Items by Category Code")
plt.xlabel("Category Code")
plt.ylabel("Number of Items")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df["Item Code"], bins=20)

plt.title("Distribution of Item Codes")
plt.xlabel("Item Code")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

## Key Insights

- The dataset contains items belonging to different categories.
- The bar chart shows the number of items available in each category.
- The pie chart represents the percentage distribution of items across categories.
- Category-wise analysis helps identify categories with a higher number of items.
- Item and category codes provide useful information for identifying and grouping products.
